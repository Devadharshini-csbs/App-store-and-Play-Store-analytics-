```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 170, 220, 250]

plt.plot(months, sales, marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.bar(months, sales, color='skyblue')
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
departments = ["HR", "IT", "Finance", "Sales"]
employees = [25, 40, 20, 15]

plt.pie(employees, labels=departments, autopct='%1.1f%%', startangle=90)
plt.title("Employee Distribution")
plt.show()
```

```python
marks = [65, 70, 75, 80, 85, 90, 95, 80, 85, 90, 75, 70, 85, 90, 95]

plt.hist(marks, bins=5, color='orange', edgecolor='black')
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
attendance = [80, 85, 90, 92, 95, 88, 84, 96]
marks_scatter = [65, 70, 80, 85, 95, 88, 72, 94]

plt.scatter(attendance, marks_scatter, color='red')
plt.title("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```

```python
salary = [25000, 27000, 30000, 32000, 35000, 40000, 150000]

plt.boxplot(salary)
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 170, 220, 250]
profit = [20, 30, 40, 35, 50, 60]

plt.plot(months, sales, label="Sales", marker='o')
plt.plot(months, profit, label="Profit", marker='s')
plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("google_play_data/googleplaystore.csv")

df.head()
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types and information:")
df.info()
```

```python
print("First 5 rows of the dataset:")
print(df.head())
```

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
print(df.columns.tolist())
```

```python
# Clean the Installs column
df['Installs'] = df['Installs'].str.replace('+', '', regex=False)
df['Installs'] = df['Installs'].str.replace(',', '', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

# Clean the Price column
df['Price'] = df['Price'].str.replace('$', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

print("Columns cleaned successfully!")
print(df[['Installs', 'Price']].head())
```

```python
print("Data types after cleaning:")
print(df[['Installs', 'Price']].dtypes)

print("\nFirst 5 cleaned records:")
print(df[['Installs', 'Price']].head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
print(df.head())
```

```python
filtered_df = df[df['Rating'] > 4.0]

print("Apps with Rating greater than 4.0:")
print(filtered_df.head(10))
```

```python
sorted_df = df.sort_values(by='Rating', ascending=False)

print("Apps sorted by Rating (highest to lowest):")
print(sorted_df[['App', 'Rating']].head(10))
```

```python
print("Rating Statistics:")

print("Mean Rating:", df['Rating'].mean())
print("Median Rating:", df['Rating'].median())
print("Minimum Rating:", df['Rating'].min())
print("Maximum Rating:", df['Rating'].max())
```

```python
category_counts = df.groupby('Category')['App'].count()

print("Number of apps in each category:")
print(category_counts)
```

```python
category_stats = df.groupby('Category')['Rating'].agg(['mean', 'min', 'max'])

print("Rating statistics by category:")
print(category_stats)
```

```python
print("Correlation Matrix:")
print(df.corr(numeric_only=True))
```

```python
print("EDA Interpretation:")
print("1. Apps with ratings greater than 4.0 were identified using filtering.")
print("2. Apps were sorted based on their ratings from highest to lowest.")
print("3. Mean, median, minimum, and maximum ratings were calculated.")
print("4. Apps were grouped according to their categories.")
print("5. Average, minimum, and maximum ratings were calculated for each category.")
print("6. Correlation between numerical variables was analyzed using a correlation matrix.")
```

```python
import matplotlib.pyplot as plt

category_counts = df['Category'].value_counts().head(10)

plt.figure(figsize=(10, 6))
category_counts.plot(kind='bar')

plt.title('Top 10 App Categories')
plt.xlabel('Category')
plt.ylabel('Number of Apps')
plt.xticks(rotation=45)
plt.show()
```

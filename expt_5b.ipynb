```python
import pandas as pd

df = pd.read_csv(r'C:\Users\Devadharshini\google_play_data\googleplaystore.csv')

print("Dataset loaded successfully!")
print(df.shape)
```

```python
import matplotlib.pyplot as plt

ratings = df['Rating'].dropna().head(20)

plt.plot(ratings)

plt.title("App Ratings - Line Chart")
plt.xlabel("App Number")
plt.ylabel("Rating")

plt.show()
```

```python
category_counts = df['Category'].value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.bar(category_counts.index, category_counts.values)

plt.title("Top 10 App Categories - Bar Chart")
plt.xlabel("Category")
plt.ylabel("Number of Apps")

plt.xticks(rotation=45)

plt.show()
```

```python
category_counts = df['Category'].value_counts().head(5)

plt.figure(figsize=(8, 8))

plt.pie(
    category_counts.values,
    labels=category_counts.index,
    autopct='%1.1f%%'
)

plt.title("Top 5 App Categories - Pie Chart")

plt.show()
```

```python
ratings = df['Rating'].dropna()

plt.figure(figsize=(8, 5))

plt.hist(ratings, bins=10)

plt.title("Distribution of App Ratings - Histogram")
plt.xlabel("Rating")
plt.ylabel("Number of Apps")

plt.show()
```

```python
df['Installs'] = df['Installs'].str.replace('+', '', regex=False)
df['Installs'] = df['Installs'].str.replace(',', '', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

print(df[['Installs', 'Rating']].head())
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Installs'], df['Rating'], alpha=0.5)

plt.title("Rating vs Number of Installs - Scatter Plot")
plt.xlabel("Number of Installs")
plt.ylabel("Rating")

plt.xscale('log')

plt.show()
```

```python
ratings = df['Rating'].dropna()

plt.figure(figsize=(8, 5))

plt.boxplot(ratings)

plt.title("App Ratings - Box Plot")
plt.ylabel("Rating")

plt.show()
```

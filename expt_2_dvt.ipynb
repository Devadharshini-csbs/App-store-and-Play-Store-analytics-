```python
import pandas as pd

# Create sample data and save as Excel
data = {'RollNo': [101, 102, 103, 104, 105],
        'Name': ['Arun', 'Bala', 'Divya', 'Priya', 'Ravi'],
        'Marks': [85, 90, 95, 88, 92]}
df = pd.DataFrame(data)
df.to_excel('students.xlsx', index=False)
print("Excel file created!")
```

```python
# Reading Excel File
df_excel = pd.read_excel('students.xlsx')
print(df_excel)
```

```python
import sqlite3
import pandas as pd

# Create SQLite database and table
conn = sqlite3.connect('student.db')
cursor = conn.cursor()
cursor.execute('CREATE TABLE IF NOT EXISTS students (RollNo INT, Name TEXT, Marks INT)')
cursor.execute('INSERT INTO students VALUES (101, "Arun", 85)')
cursor.execute('INSERT INTO students VALUES (102, "Bala", 90)')
cursor.execute('INSERT INTO students VALUES (103, "Divya", 95)')
conn.commit()

# Read from SQL
df_sql = pd.read_sql("SELECT * FROM students", conn)
print(df_sql)
conn.close()
```

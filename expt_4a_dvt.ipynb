```python
import pandas as pd
import numpy as np

# Student Performance Dataset
data = {
    'Roll No': [101, 102, 103, 104, 105, 106],
    'Name': ['Arun', 'Bala', 'Divya', 'Priya', 'Ravi', 'Karthik'],
    'Department': ['CSE', 'IT', 'CSE', 'AI&DS', 'IT', 'AI&DS'],
    'Marks': [85, 78, 95, 90, 72, 88],
    'Attendance': [92, 88, 96, 94, 85, 91]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Display first 5 records
print(df.head())
```

```python
# Shape of dataset
print(df.shape)
```

```python
# Data types
print(df.dtypes)
```

```python
# Students scoring more than 85
high_marks = df[df["Marks"] > 85]
print(high_marks)
```

```python
# Students from IT Department
it_students = df[df["Department"] == "IT"]
print(it_students)
```

```python
# Sort Marks in Descending Order
sorted_df = df.sort_values(by="Marks", ascending=False)
print(sorted_df)
```

```python
print("Mean =", df['Marks'].mean())
print("Median =", df['Marks'].median())
print("Mode =", df['Marks'].mode()[0] if not df['Marks'].mode().empty else "No Repetition")
print("Standard Deviation =", df['Marks'].std())
```

```python
print(df.describe())
```

```python
dept_avg = df.groupby("Department")["Marks"].mean()
print(dept_avg)
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['Attendance'], df['Marks'])
plt.title("Marks vs Attendance")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```
